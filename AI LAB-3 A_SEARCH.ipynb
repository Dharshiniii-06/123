```python
import heapq

def a_star_search(graph, heuristic, initial_state, goal_state):
    initial_node = {
        "state": initial_state,
        "path": [initial_state],
        "g": 0,
        "h": heuristic[initial_state],
        "f": heuristic[initial_state]
    }

    if initial_state == goal_state:
        return initial_node["path"]

    frontier = []
    counter = 0

    heapq.heappush(
        frontier,
        (initial_node["f"], counter, initial_node)
    )

    explored = set()

    while frontier:
        _, _, node = heapq.heappop(frontier)

        current_state = node["state"]
        current_path = node["path"]
        current_g = node["g"]
        current_h = node["h"]
        current_f = node["f"]

        if current_state in explored:
            continue

        print(f"\nExpanding node: {current_state}")
        print(f"Current path: {current_path}")
        print(
            f"g={current_g}, h={current_h}, "
            f"f={current_f}"
        )

        if current_state == goal_state:
            print(f"\nGoal state {goal_state} found!")
            return current_path

        explored.add(current_state)

        frontier_states = {
            frontier_node["state"]
            for _, _, frontier_node in frontier
        }

        for child_state, cost in graph.get(current_state, []):
            if (child_state not in explored
                    and child_state not in frontier_states):

                child_g = current_g + cost
                child_h = heuristic[child_state]
                child_f = child_g + child_h

                child_node = {
                    "state": child_state,
                    "path": current_path + [child_state],
                    "g": child_g,
                    "h": child_h,
                    "f": child_f
                }

                print(
                    f"Generated child: {child_state}, "
                    f"g={child_g}, h={child_h}, f={child_f}"
                )

                counter += 1

                heapq.heappush(
                    frontier,
                    (child_f, counter, child_node)
                )

        print(
            "Frontier:",
            [
                (frontier_node["state"], frontier_node["f"])
                for _, _, frontier_node in frontier
            ]
        )

        print("Explored:", explored)

    return None


graph = {
    "A": [("B", 2), ("C", 4)],
    "B": [("A", 2), ("D", 3), ("E", 5)],
    "C": [("A", 4), ("F", 2), ("G", 4)],
    "D": [("B", 3)],
    "E": [("B", 5), ("H", 3)],
    "F": [("C", 2)],
    "G": [("C", 4), ("I", 2)],
    "H": [("E", 3)],
    "I": [("G", 2)]
}

heuristic = {
    "A": 6,
    "B": 5,
    "C": 4,
    "D": 7,
    "E": 4,
    "F": 4,
    "G": 2,
    "H": 3,
    "I": 0
}

initial_state = "A"
goal_state = "I"

solution = a_star_search(
    graph, heuristic, initial_state, goal_state
)

if solution is not None:
    print("\nSolution path:")
    print(" -> ".join(solution))
else:
    print("\nFailure: No path exists.")
```
