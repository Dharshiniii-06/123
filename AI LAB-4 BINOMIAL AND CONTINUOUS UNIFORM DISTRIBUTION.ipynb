```python
#Binomial Distribution
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parameters
n = 8 # No. of trials
p = 0.5 # Probability of success

x = np.arange(n + 1)

probability = binom.pmf(x, n, p)

plt.bar(x, probability, alpha=0.5, label="PMF")
plt.plot(x, probability, marker="o", label="PMF line")

plt.title("Binomial Distribution")
plt.xlabel("Number of successes")
plt.ylabel("Probability")
plt.xticks(x)
plt.legend()
plt.grid(axis="y")
plt.show()
```

```python
#Continuous Distribution
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform

x1 = uniform.rvs(loc=0.01, scale=0.09, size=100)
x2 = uniform.rvs(loc=0.01, scale=0.99, size=100)

pdf_value = uniform.pdf(x1[0], loc=0.01, scale=0.99)
print("PDF value:", pdf_value)

plt.hist(x1, bins=10, density=True, alpha=0.7, label="Small")
plt.hist(x2, bins=10, density=True, alpha=0.7, label="Large")

plt.axhline(y=1/0.09, color="blue", linewidth=2, label="Small PDF")
plt.axhline(y=1/0.99, color="black", linewidth=2, label="Large PDF")

plt.title("Continuous Uniform Distribution")
plt.xlabel("x")
plt.ylabel("Density")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform


x = uniform.rvs(loc=0.01, scale=0.99, size=100)

# PDF value at the first generated point
pdf_value = uniform.pdf(x[0], loc=0.01, scale=0.99)

print("PDF value =", pdf_value)

plt.hist(x, bins=10, density=True, alpha=0.7, label="Generated data")

plt.axhline(y=1 / 0.99, color="black",
            linewidth=2, label="Theoretical PDF")

plt.title("Continuous Uniform Distribution")
plt.xlabel("x")
plt.ylabel("Density")
plt.legend()

plt.show()
```
