```python
import heapq

def ucs(graph, start, goal):
    pq = [(0, [start])]
    explored = set()

    while pq:
        cost, path = heapq.heappop(pq)
        node = path[-1]

        if node in explored:
            continue

        print("\n Expanding node:", node)
        print("Current path:", path)
        print("Path cost:", cost)

        if node == goal:
            print("\n Goal state", goal, "found!")
            return path, cost

        explored.add(node)

        for child, step_cost in graph[node]:
            if child not in explored:
                new_path = path + [child]
                new_cost = cost + step_cost

                print("Generated child:", child)
                print("New path:", new_path)
                print("New cost:", new_cost)

                heapq.heappush(pq, (new_cost, new_path))

        print("Frontier:", [(p[-1], c) for c, p in pq])
        print("Explored:", explored)

    return None, None


graph = {
    'A': [('E', 1), ('F', 2), ('G', 3)],
    'B': [('E', 2)],
    'C': [('F', 1)],
    'D': [('F', 4)],
    'E': [('A', 1), ('B', 2)],
    'F': [('A', 2), ('C', 1), ('D', 4)],
    'G': [('A', 3), ('H', 2), ('I', 3)],
    'H': [('G', 2), ('I', 1)],
    'I': [('G', 3), ('H', 1)]
}


solution, cost = ucs(graph, 'A', 'I')

if solution:
    print("\n Solution path:")
    print("-->".join(solution))
    print("Path cost:", cost)
else:
    print("\n Failure: no path exists")
```
