```python
#Discrete Uniform Distribution
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import randint

low = 1
high = 7

x = np.arange(low, high)
probability = randint.pmf(x, low, high)

plt.bar(x, probability, alpha=0.7, label="PMF")
plt.title("Discrete Uniform Distribution")
plt.xlabel("x")
plt.ylabel("Probability")
plt.xticks(x)
plt.legend()
plt.show()
```

```python
#Bernoulli Distribution
import numpy as np
import matplotlib.pyplot as plt

p = 0.3
x = np.array([0, 1])

probability = (p**x) * (1-p)**(1-x)

print("Possible outcomes:", x)
print("Probabilities:", probability)

plt.bar(x, probability, alpha=0.7, label="PMF")
plt.title("Bernoulli Distribution")
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.xticks(x, ["Failure (0)", "Success (1)"])
plt.legend()
plt.show()
```

```python
# Bernoulli Distribution
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli

# Probability of success
p = 0.6

# Generate 100 random values
x = bernoulli.rvs(p, size=100)

# Calculate PMF
probability = bernoulli.pmf([0, 1], p)

print("PMF values:", probability)

# Plot bar graph
plt.bar([0, 1], probability, alpha=0.7, label="PMF")

plt.title("Bernoulli Distribution")
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.xticks([0, 1], ["Failure (0)", "Success (1)"])
plt.legend()
plt.grid(axis="y")

plt.show()
```
